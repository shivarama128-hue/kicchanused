```python

import cv2
from pathlib import Path
from google.colab.patches import cv2_imshow

# Find our saved image
image_path = Path("student_faces/S001_Ravi/sample.jpg")

# Read the image
image = cv2.imread(str(image_path))

if image is None:
    raise FileNotFoundError("Image not found. Check the file path.")

# Load OpenCV's face detector
model_url = (
    "https://raw.githubusercontent.com/opencv/opencv/master/"
    "data/haarcascades/haarcascade_frontalface_default.xml"
)

import requests

response = requests.get(model_url, timeout=30)
response.raise_for_status()

model_path = "haarcascade_frontalface_default.xml"

with open(model_path, "wb") as file:
    file.write(response.content)

face_detector = cv2.CascadeClassifier(model_path)

# Convert the picture to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Find faces
faces = face_detector.detectMultiScale(
    gray_image,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(30, 30)
)

# Draw a rectangle around every detected face
for (x, y, w, h) in faces:
    cv2.rectangle(
        image,
        (x, y),
        (x + w, y + h),
        (0, 255, 0),
        3
    )

print("Number of faces detected:", len(faces))
cv2_imshow(image)
```

```python
import cv2
import requests
from pathlib import Path
from google.colab.patches import cv2_imshow

# Step 1: Download a sample face image
image_url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
response = requests.get(image_url, timeout=30)
response.raise_for_status()

image_path = Path("sample.jpg")
image_path.write_bytes(response.content)

# Step 2: Load the image
image = cv2.imread(str(image_path))

if image is None:
    raise ValueError("Image could not be loaded!")

# Step 3: Download the face-detection model
model_url = "https://raw.githubusercontent.com/opencv/opencv/master/data/haarcascades/haarcascade_frontalface_default.xml"
response = requests.get(model_url, timeout=30)
response.raise_for_status()

model_path = "haarcascade_frontalface_default.xml"
Path(model_path).write_bytes(response.content)

# Step 4: Load the model
face_detector = cv2.CascadeClassifier(model_path)

if face_detector.empty():
    raise ValueError("Face-detection model could not be loaded!")

# Step 5: Detect the face
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

faces = face_detector.detectMultiScale(
    gray_image,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(30, 30)
)

# Step 6: Draw a green rectangle around each face
for (x, y, w, h) in faces:
    cv2.rectangle(
        image,
        (x, y),
        (x + w, y + h),
        (0, 255, 0),
        3
    )

print("Number of faces detected:", len(faces))
cv2_imshow(image)
```

```python
import cv2
from pathlib import Path
from google.colab.patches import cv2_imshow

# Open Ravi's sample image
image_path = Path("student_faces/S001_Ravi/sample.jpg")
image = cv2.imread(str(image_path))

# Convert the image to gray
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Load the face detector
detector = cv2.CascadeClassifier(
    "haarcascade_frontalface_default.xml"
)

# Find faces
faces = detector.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(30, 30)
)

if len(faces) == 0:
    print("No face found. Try the previous detection step again.")
else:
    # Take the first detected face
    x, y, w, h = faces[0]

    # Cut out the face
    face = image[y:y+h, x:x+w]

    # Save the cropped face
    face_path = Path("student_faces/S001_Ravi/face.jpg")
    cv2.imwrite(str(face_path), face)

    print("Face saved successfully!")
    cv2_imshow(face)
```

```python
# @title AI prompt cell

import ipywidgets as widgets
from IPython.display import display, HTML, Markdown,clear_output
from google.colab import ai

dropdown = widgets.Dropdown(
    options=[],
    layout={'width': 'auto'}
)

def update_model_list(new_options):
    dropdown.options = new_options
update_model_list(ai.list_models())

text_input = widgets.Textarea(
    placeholder='Ask me anything....',
    layout={'width': 'auto', 'height': '100px'},
)

button = widgets.Button(
    description='Submit Text',
    disabled=False,
    tooltip='Click to submit the text',
    icon='check'
)

output_area = widgets.Output(
     layout={'width': 'auto', 'max_height': '300px','overflow_y': 'scroll'}
)

def on_button_clicked(b):
    with output_area:
        output_area.clear_output(wait=False)
        accumulated_content = ""
        for new_chunk in ai.generate_text(prompt=text_input.value, model_name=dropdown.value, stream=True):
            if new_chunk is None:
                continue
            accumulated_content += new_chunk
            clear_output(wait=True)
            display(Markdown(accumulated_content))

button.on_click(on_button_clicked)
vbox = widgets.GridBox([dropdown, text_input, button, output_area])

display(HTML("""
<style>
.widget-dropdown select {
    font-size: 18px;
    font-family: "Arial", sans-serif;
}
.widget-textarea textarea {
    font-size: 18px;
    font-family: "Arial", sans-serif;
}
</style>
"""))
display(vbox)
```
